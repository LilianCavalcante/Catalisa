```python
# 📝 ATIVIDADE 2: Cadastro de Professor
# Desenvolva um programa que receba o nome de um professor, seu salário fixo e o total de aulas mensais.

nome = input("Digite o nome do professor?")
salario_fixo = float(input("Qual o salário fixo?"))
aulas_mensais = int(input("Qual o total de aulas mensais?"))

print(f"O nome do professor é: {nome}")
print(f"O salário fixo é de: R$ {salario_fixo}")
print(f"O total de aulas mensais é de: {aulas_mensais}")
```

```python
# 📝 ATIVIDADE 3: Cálculo do Delta (Equação do 2º Grau)
# Leia os coeficientes A, B e C e exiba o valor de Delta (Δ = b² - 4ac)

a = float(input("Digite o valor de A: "))
b = float(input("Digite o valor de B: "))
c = float(input("Digite o valor de C: "))

delta = b**2 - (4*a*c)
print(f"O valor de Delta é: {delta}")

```

```python
# 📝 ATIVIDADE 4: Mini Calculadora Aritmética
# Receba dois números e exiba as 4 operações básicas aplicadas a eles.

a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))

soma = a + b
subtracao = a - b
divisao = a / b
multiplicacao = a * b

print(f"Soma é = {soma}")
print(f"Subtração é = {subtracao}")
print(f"Divisão é = {divisao}")
print(f"Multiplicação é = {multiplicacao}")

```

```python
# 📝 ATIVIDADE 5: Operadores Relacionais
# Verifique de forma direta se o número A é maior do que o número B.

a = float(input("Digite o número A: "))
b = float(input("Digite o número B: "))

if a > b:
  print(f"O número A é maior que o número B")

else:
    print(f"O número A é menor que o número B")
```
